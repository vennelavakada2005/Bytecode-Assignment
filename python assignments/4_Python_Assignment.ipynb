## 6.Find the average (marks = [70, 80, 90, 60])

```python
marks=[30,40,50,70,20]
total=0
for i in marks:
    total=total+i
    avg=total/len(marks)
print(total)
print(avg)
```

## 11.Count vowels

```python
ls="hello python"
vowels="aeiouAEIOU"
count=0
for i in ls:
    if i in vowels:
        count=count+1
print(count)
```

## 12.Count uppercase letters

```python
ls="HELLO PYTHON"
count=0
for i in ls:
    if i.isupper():
        count=count+1
print(count)
```

## 20.calculate total sales

```python
sales={"shirts":100,
       "pants":200,
       "shoes":300}
total_sales=0
for i in sales.values():
    total_sales=total_sales+i
print(total_sales)
```

## 22.Count customers with purchases above 1000

```python
sales={"shirts":100,
       "pants":200,
       "shoes":3000}
count=0
for i in sales.values():
    if i>1000:
        count=count+1
print(count)

```
