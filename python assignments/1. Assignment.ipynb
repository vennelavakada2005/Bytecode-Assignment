Q7. Sort the following list. numbers = [50, 20, 80, 10, 40]
Reverse sort the list.


```python
L1=[50,20,80,10,40]
print("Before sort:",L1)
L1.sort() #ASC
print("After sort ASC:",L1)
L1.sort(reverse = True) #DESC
print("After sort DESC:",L1)
```
