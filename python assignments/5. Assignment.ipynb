## 16.	Second Highest Sale
sales = [1200, 3000, 1800, 5000, 4500]
Find the second highest sale using a while loop.


```python
sales = [1200, 3000, 1800, 5000, 4500]

highest = sales[0]
second_highest = sales[0]

i = 1

while i < len(sales):
    if sales[i] > highest:
        second_highest = highest
        highest = sales[i]
    elif sales[i] > second_highest and sales[i] != highest:
        second_highest = sales[i]
    i += 1

print("Second highest sale:", second_highest)
```

## 21.	Build Frequency Dictionary
products = [“Laptop”, “Mobile”, “Laptop”, “Tablet”, “Mobile”, “Laptop”]
Create a frequency dictionary using a while loop.


```python
products = ["Laptop", "Mobile", "Laptop", "Tablet", "Mobile", "Laptop"]

frequency = {}
i = 0

while i < len(products):
    product = products[i]

    if product in frequency:
        frequency[product] += 1
    else:
        frequency[product] = 1

    i += 1

print(frequency)
```

## 25.	Mini KPI Dashboard
sales = [1200, 1500, 1800, 2000, 1700]
Using a single while loop, calculate:
- 	Total Sales
- 	Average Sales
- 	Maximum Sales
- 	Minimum Sales
- 	Number of Records


```python
sales = [1200, 1500, 1800, 2000, 1700]

total_sales = 0
max_sales = sales[0]
min_sales = sales[0]

i = 0

while i < len(sales):
    total_sales += sales[i]

    if sales[i] > max_sales:
        max_sales = sales[i]

    if sales[i] < min_sales:
        min_sales = sales[i]

    i += 1

num_records = len(sales)
average_sales = total_sales / num_records

print("Total Sales:", total_sales)
print("Average Sales:", average_sales)
print("Maximum Sales:", max_sales)
print("Minimum Sales:", min_sales)
print("Number of Records:", num_records)
```

## 30.	Sales Trend Analysis
sales = [1000, 1200, 1100, 1500, 1400]
Count:
•	Number of increasing months
•	Number of decreasing months.


```python
sales = [1000, 1200, 1100, 1500, 1400]

increasing = 0
decreasing = 0

i = 1

while i < len(sales):
    if sales[i] > sales[i - 1]:
        increasing += 1
    elif sales[i] < sales[i - 1]:
        decreasing += 1

    i += 1

print("Number of increasing months:", increasing)
print("Number of decreasing months:", decreasing)
```
