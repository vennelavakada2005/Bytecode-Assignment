## data = ["100", "200", "300", "400"]
Convert all values to integers and find:
- Total sum
- Average
- Maximum value
- Minimum value


```python
data = ["100", "200", "300", "400"]

# Convert strings to integers
numbers = [int(x) for x in data]

total_sum = sum(numbers)
average = total_sum / len(numbers)
maximum = max(numbers)
minimum = min(numbers)

print("Total Sum:", total_sum)
print("Average:", average)
print("Maximum Value:", maximum)
print("Minimum Value:", minimum)
```

## ■ Question 7
text = "Data123Analytics456Python789"
Extract:
- Only alphabets
- Only digits
- Count total digits
- Convert extracted text into uppercase


```python
text = "Data123Analytics456Python789"

# Extract only alphabets
alphabets = ""
for ch in text:
    if ch.isalpha():
        alphabets += ch
# Extract only digits
digits = ""
for ch in text:
    if ch.isdigit():
        digits += ch
# Count total digits
total_digits = len(digits)
# Convert extracted text into uppercase
uppercase_text = alphabets.upper()
print("Only Alphabets:", alphabets)
print("Only Digits:", digits)
print("Total Digits:", total_digits)
print("Uppercase Text:", uppercase_text)
```

## ★ Challenge 2
students = {
"Ram":[80,90,85],
"Priya":[95,88,92],
"Hari":[70,75,80]
}
Find:
- Total marks of each student
- Average marks of each student
- Student with highest average


```python
students = {
    "Ram": [80, 90, 85],
    "Priya": [95, 88, 92],
    "Hari": [70, 75, 80]
}

averages = {}
# Total of each student
for name, marks in students.items():
    total = sum(marks)
#average of each student
    average = total / len(marks)
    averages[name] = average
    print(f"{name} Total Marks: {total}")
    print(f"{name} Average Marks: {average:.2f}")
# Student with highest average
highest_avg_student = max(averages, key=averages.get)
print("\nStudent with Highest Average:", highest_avg_student)
print("Highest Average:", averages[highest_avg_student])
```

## ★ Challenge 3
data = ["Python","SQL","Excel","Python","Power BI","SQL","Python"]
Find:
- Unique technologies
- Frequency of each technology
- Technology with highest frequency


```python
data = ["Python", "SQL", "Excel", "Python", "Power BI", "SQL", "Python"]

# Unique technologies
unique_tech = set(data)
print("Unique Technologies:", unique_tech)
# Frequency of each technology
frequency = {}
for tech in data:
    frequency[tech] = frequency.get(tech, 0) + 1
print("Frequency of Each Technology:", frequency)
# Technology with highest frequency
highest_freq_tech = max(frequency, key=frequency.get)
print("Technology with Highest Frequency:", highest_freq_tech)
print("Frequency:", frequency[highest_freq_tech])
```
