7. Data Cleaning : sales = [100,200,100,300,200,400,500,400]
Tasks:
- Remove duplicates using sets.
-	Find count of unique sales values.
-	Convert result back to list.


```python
sales = [100,200,100,300,200,400,500,400]
unique_sales=set(sales)
print("unique sales:",unique_sales)

count=len(unique_sales)
print("count of unique sales values:",count)

sales_list=list(unique_sales)
print("converted to list:",sales_list)
```

```python
day1 = {101,102,103,104,105}
day2 = {103,104,105,106,107}
day3 = {104,105,108}
print("all 3 days:",day1&day2&day3)
print("at least one:",day1|day2|day3)
print("only day 1:",day1-day2-day3)
```

10.Duplicate Customer IDs : customers = [101,102,103,101,104,105,102,106]
Tasks:
- Find unique customers
-	Count duplicates removed
-	Find total unique customers



```python
customers = [101,102,103,101,104,105,102,106]
unique_customers=set(customers)
print("unique_customers:",unique_customers)

duplicates_removed=len(customers)-len(unique_customers)
print("duplicates removed:",duplicates_removed)

total_unique=len(unique_customers)
print("total unique customers:",total_unique)
```

13. Mini Case Study
An e-commerce company ran campaigns in 3 months:

jan = {101,102,103,104,105}
feb = {103,104,105,106,107}
mar = {105,106,107,108,109}

Answer:

-	Total unique customers targeted.
-	Customers targeted in all months.
-	Customers targeted only once.
-	Customers targeted in January and February but not March.
-	Customers targeted in February and March but not January.


```python
jan = {101,102,103,104,105}
feb = {103,104,105,106,107}
mar = {105,106,107,108,109}
print("total unique customers:",jan|feb|mar)
print("all months:",jan&feb&mar)

only_once=(jan - feb - mar) | (feb-mar-jan)|(mar-jan-feb)
print("only once:",only_once)

print("jan&feb not mar:",(jan&feb)-mar)
print("feb&mar not jan:",(feb&mar)-jan)
```
